#Week 4: Preparing datafor machine learning
This week we are turning "clean" data into "mathematical" data
Tasks:
1. Encoding categorical feature(text -> numbers).
2. Normalization(converting numbers to the same scale).
3. Splitting the data into sets: Training(Train), Validation(Val), and Test(Test).

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

# Let's repeat the quick cleanse from Week 3
df = df.drop(['Ticket',"Cabin","Name"], axis = 1)
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

print("Данные загружены. Формат:", df.shape)
```

```python
# 1. Encoding gender and landing port (Categorical to Numerical)
df = pd.get_dummies(df, columns = ['Sex', 'Embarked'], drop_first = True)

# 2. Scaling (Normalization)
# To ensure that age (0-80) and ticket price (0-500) are given equal weight in the model
scaler = StandardScaler()
df[["Age", "Fare"]] = scaler.fit_transform(df[['Age', 'Fare']])

print("The data is encoded and normalized.")
df.head()
```

```python
import os

def split_and_save(dataframe, output_dir = 'data'):
  if not os.path.exists(output_dir):
    os.makedirs(output_dir)

  # First we separate the Test (15%)
  train_val, test = train_test_split(dataframe, test_size = 0.15, random_state = 42)

  # Then we divide the remainder into Train and Val (approximately 15% of the total volume)
  train, val = train_test_split(train_val, test_size = 0.18, random_state = 42)

  # Save
  train.to_csv(f'{output_dir}/train.csv', index = False)
  val.to_csv(f'{output_dir}/val.csv', index = False)
  test.to_csv(f'{output_dir}/test.csv', index = False)

  print(f"Files saved to folder {output_dir}:")
  print(f"- Train: {train.shape[0]} lines")
  print(f"- Val: {val.shape[0]} lines")
  print(f"- Test: {test.shape[0]} lines")

# Running the function
split_and_save(df)
```
