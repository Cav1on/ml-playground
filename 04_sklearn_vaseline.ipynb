# Week 5: Classical machine learning (Baseline)
1. Train Logistic Regression.
2. Train Random Forest.
3. Use Cross-Validation for fair evalution.

```python
import pandas as pd
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler

# 1. Loading and quick cleaning
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
df = df.drop(['Ticket', 'Cabin', 'Name', 'PassengerId'], axis = 1)
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# 2. Encoding and scaling
df = pd.get_dummies(df, columns = ['Sex', 'Embarked'], drop_first = True)
scaler = StandardScaler()
df[['Age','Fare']] = scaler.fit_transform(df[['Age', 'Fare']])

# 3. Separation into features (x) and goal (y)
x = df.drop('Survived', axis = 1)
y = df['Survived']

# We divide the data into training and test sets.
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)

print("The data is ready for training!")
```

```python
log_reg = LogisticRegression()
log_reg.fit(x_train, y_train)

# We check the accuracy on test data
accuracy_log = log_reg.score(x_test, y_test)
print(f"Accuracy of Logistic Regression: {accuracy_log:.4f}")
```

```python
rf_model = RandomForestClassifier(n_estimators = 100, random_state = 42)
rf_model.fit(x_train, y_train)

accuracy_rf = rf_model.score(x_test, y_test)
print(f"Accuracy of Random Forest: {accuracy_rf:.4f}")
```

```python
# Testing Random Forest on 5 different folds (parts of the data)
cv_scores = cross_val_score(rf_model, x, y, cv = 5)

print(f"Average accuracy in cross-validation: {cv_scores.mean():.4f}")
print(f"Accuracy Variation: {cv_scores.std():.4f}")
```
